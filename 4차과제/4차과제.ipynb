# 서울대공원 혼잡 예측 서비스 분석 보고서

## 1. 서비스 정의

이 서비스는 서울대공원에 방문하려는 사람이 **어느 날이 덜 붐빌지 미리 확인할 수 있게 도와주는 서비스**입니다.

정확히 몇 명이 방문할지 맞히는 것보다, 사용자가 쉽게 이해할 수 있도록 날짜별로 **한산함, 보통, 붐빔, 매우 붐빔**처럼 혼잡 정도를 알려주는 것이 핵심입니다.

## 2. 주요 타겟층

| 구분 | 타겟층 | 필요한 점 |
|---|---|---|
| 주요 타겟 | 아이와 함께 방문하는 가족 | 주차, 대기, 이동 피로를 줄이고 싶어 합니다 |
| 주요 타겟 | 주말 나들이 방문객 | 덜 붐비는 날짜를 고르고 싶어 합니다 |
| 보조 타겟 | 연인·친구 방문객 | 날씨, 축제, 혼잡도를 함께 보고 싶어 합니다 |
| 보조 타겟 | 단체 방문 기획자 | 안전하고 덜 붐비는 방문일을 찾고 싶어 합니다 |
| 보조 타겟 | 서울대공원 운영자 | 인력, 주차, 안내 준비를 미리 하고 싶어 합니다 |

가장 중요한 타겟은 **아이와 함께 방문하는 가족**입니다. 사람이 많으면 주차, 이동, 식사, 휴식 모두 어려워지기 때문입니다.

## 3. 실제 사용자 사용 예상 시나리오

### 시나리오 1. 아이와 함께 주말 방문을 준비하는 가족

부모님은 주말에 아이와 서울대공원에 가려고 합니다. 하지만 너무 붐비면 주차가 어렵고, 아이가 오래 기다리기 힘들 수 있습니다.

이때 서비스에서 이번 주 토요일과 일요일의 혼잡도를 비교합니다. 토요일이 `매우 붐빔`, 일요일이 `보통`으로 나오면 일요일 방문을 선택할 수 있습니다.

이 사용자는 정확한 방문객 수보다 **“우리 가족이 편하게 다녀올 수 있는 날인가?”**를 알고 싶어 합니다.

### 시나리오 2. 축제 기간에 방문하려는 연인 또는 친구

사용자는 장미축제나 벚꽃 시즌에 서울대공원에 가고 싶어 합니다. 하지만 축제 기간에는 사람이 많을 가능성이 큽니다.

서비스에서 축제 기간 중 혼잡도가 낮은 날짜를 확인하고, 날씨까지 함께 봅니다. 예를 들어 평일이고 비가 오지 않으며 `보통`으로 예측되는 날을 고를 수 있습니다.

이 사용자는 **축제는 즐기고 싶지만 너무 붐비는 날은 피하고 싶어 합니다.**

### 시나리오 3. 어린이집이나 학교에서 단체 방문일을 정하는 경우

단체 방문 담당자는 여러 후보 날짜 중 안전하고 이동이 쉬운 날을 골라야 합니다.

서비스에서 후보 날짜 3~5개를 비교합니다. `붐빔`이나 `매우 붐빔`인 날은 피하고, `한산함` 또는 `보통`인 날을 선택합니다.

이 사용자는 **혼잡도가 낮고 날씨가 안정적인 날짜**를 찾고 싶어 합니다.

### 시나리오 4. 당일 아침에 방문 여부를 결정하는 사람

사용자는 아침에 날씨를 보고 서울대공원에 갈지 말지 결정하려고 합니다.

서비스에서 오늘의 혼잡도와 날씨 정보를 확인합니다. `보통`이고 비가 거의 없다면 방문을 결정할 수 있습니다. 반대로 `매우 붐빔`이고 축제 기간이라면 다른 날짜를 생각할 수 있습니다.

이 사용자는 **빠르게 오늘 가도 괜찮은지 판단**하고 싶어 합니다.

### 시나리오 5. 서울대공원 운영자가 주말 준비를 하는 경우

운영자는 이번 주말에 사람이 많이 올지 미리 알고 싶어 합니다.

서비스에서 주말 혼잡도가 높게 나오면 주차 안내, 입장 안내, 안전 관리, 인력 배치를 더 준비할 수 있습니다.

이 사용자는 방문객 개인이 아니라 **운영 준비를 더 잘하기 위해 예측 결과를 사용**합니다.

| 데이터 | 쓰임 | 안정성 | 판단 |

## 5. 예측 실패 리스크

예측 실패 중 가장 위험한 경우는 **실제로는 많이 붐비는데, 서비스가 한산하다고 알려주는 경우**입니다.

| 실패 유형 | 생길 수 있는 문제 | 위험도 |
|---|---|---:|
| 붐비는 날을 낮게 예측 | 주차 실패, 긴 대기, 고객 불만 | 매우 높음 |
| 한산한 날을 높게 예측 | 사용자가 방문을 포기할 수 있음 | 중간 |
| 방문객 수는 틀렸지만 등급은 맞음 | 큰 문제는 적음 | 낮음 |
| 축제 정보를 빠뜨림 | 사람이 많은 날을 놓칠 수 있음 | 높음 |
| 날씨가 갑자기 바뀜 | 단기 예측이 틀릴 수 있음 | 중간~높음 |

따라서 이 모델은 평균적으로 숫자를 잘 맞히는 것보다 **붐비는 날을 놓치지 않는 것**이 더 중요합니다.

이번 수정의 핵심은 **서비스를 실제 사용자가 더 쉽게 이해하고 판단할 수 있게 만든 것**입니다.

모델 입력은 실제 미래 예측에 맞게 정리되었고, 화면에서는 공휴일, 행사 시즌, 혼잡도, 위험 색상이 더 잘 보이도록 개선되었습니다. 또한 예측 결과는 단순한 방문객 수가 아니라, 사용자가 방문 준비에 바로 활용할 수 있는 쉬운 안내 문장으로 바뀌었습니다.
